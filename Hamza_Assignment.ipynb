```python

movie_clean = movie.copy()
movie_clean["Runtime"] = movie_clean["Runtime"].str.replace(" min","", regex=False).astype(float)
movie_clean["Runtime"].dtype
```

```python

movie_clean["Gross"] = movie_clean["Gross"].fillna(movie_clean["Gross"].median())
movie_clean["Gross"].isna().sum()
```
