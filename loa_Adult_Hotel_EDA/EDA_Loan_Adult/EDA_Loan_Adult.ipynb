```python
adult.replace('?', np.nan, inplace=True)
```

```python
adult.isnull().sum()
```

```python
adult.isnull().sum() / len(adult)*100
```

```python
print("Duplicate rows:", adult.duplicated().sum())
adult = adult.drop_duplicates()
```

```python
adult['workclass'] = adult['workclass'].fillna(adult['workclass'].mode()[0])
```

```python
adult['occupation'] = adult['occupation'].fillna(adult['occupation'].mode()[0])
```

```python
adult['native-country'] = adult['native-country'].fillna(adult['native-country'].mode()[0])
```

```python
print(adult.isnull().sum())
```

```python
print(adult.dtypes)
```

```python
numeric_cols = ['age', 'fnlwgt', 'educational-num', 'capital-gain', 'capital-loss', 'hours-per-week']
```

```python
categorical_cols = ['workclass', 'education', 'marital-status', 'occupation', 'relationship', 'race', 'gender', 'native-country', 'income']
```

```python
for col in numeric_cols:
    plt.figure(figsize=(12,4))

    plt.subplot(1,2,1)
    sns.histplot(adult[col], kde=True, bins=30)
    plt.title(f'Histogram of {col}')

    plt.subplot(1,2,2)
    sns.boxplot(x=adult[col])
    plt.title(f'Boxplot of {col}')

    plt.tight_layout()
    plt.show()
```

```python
for col in categorical_cols:
    print(adult[col].value_counts())
    plt.figure(figsize=(8,4))
    sns.countplot(y=col, data=adult, order=adult[col].value_counts().index, palette='Set2')
    plt.title(f'Distrbution of {col}')
    plt.show()
```

```python
for col in ['age', 'fnlwgt', 'hours-per-week']:
    Q1 = adult[col].quantile(0.25)
    Q3 = adult[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 - 1.5 * IQR

    adult[col] = np.where(adult[col] > upper_bound, upper_bound, adult[col])
    adult[col] = np.where(adult[col] < lower_bound, lower_bound, adult[col])
```

```python
for col in  ['age', 'fnlwgt', 'hours-per-week']:
    plt.figure(figsize=(5,4))
    sns.boxplot(x=adult[col])
    plt.title(f'Boxplot after Outlier Treatment - {col}')
    plt.show()
```

```python
for col in ['workclass', 'education', 'marital-status', 'occupation', 'relationship', 'race', 'gender']:
    plt.figure(figsize=(8,4))
    sns.countplot(y=col, hue='income', data=adult, order=adult[col].value_counts().index, palette="Set1")
    plt.title(f"{col} vs income")
    plt.show()
```

```python

for col in numeric_cols:
    plt.figure(figsize=(6,4))
    sns.boxplot(x='income', y=col, data=adult, palette="Set3")
    plt.title(f"{col} vs income")
    plt.show()
```

```python
plt.figure(figsize=(10,7))
corr = adult[numeric_cols].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap - Adult Dataset")
plt.show()
```
